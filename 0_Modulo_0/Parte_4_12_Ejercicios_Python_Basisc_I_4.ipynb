## Ejercicio 2
Para el siguiente string se pide imprimir por pantalla los siguientes casos

    "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."

1. Pásalo todo a mayúsculas
2. Pásalo todo a minúsculas
3. Solo la primera letra de cada palabra en mayúscula, el resto en minúscula
4. Crea una lista dividiéndolo por sus espacios
5. Sustituye las comas `,` por puntos y comas `;`
6. Elimina las `a` minúsculas

```python
#1
frase = "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."
print(frase.upper())
print(frase.lower())

print(frase.title())  # primera letra en mayuscula, resto en minuscula

print(frase.split(" "))  # lista dividida por sus espacios 
print(frase.replace(",", ";"))
print(frase.replace("a", ""))

# split significa “dividir”. 
# Corta el texto cada vez que encuentra el separador que le indicas y devuelve una lista con los trozos.
```

## Ejercicio 3
1. Crea una lista con 3 elementos numéricos
2. Añade un cuarto elemento
3. Calcula la suma de todos
4. Elimina el segundo elemento de la lista
5. Añade otro elemento en la posicion 3 de la lista
6. Crea otra lista con 4 elementos y concaténala a la que ya tenías.
7. Ordena la lista de menor a mayor.
8. BONUS: Ordena la lista de mayor a menor.  
(*PISTA: Para hacer 8, tendrás que buscar en la información del método...*)  
  
  
NOTA: Pinta por pantalla el resultado de cada paso.

```python
lista1 = [3, 2, 5]
print(lista1)

lista1.append(9)
print(lista1)

print(sum(lista1))

print(lista1.pop(1))   # sale en print 2 el elemento eliminado
print(lista1)  # sale en print la lista sin el elemento eliminado el 2

# añadir otro elemento en la posicion 3 de la lista, en este caso el numero 7
lista1.insert(2,7)
print(lista1)  

lista2 = ["arbol", 4, "casa", -1]
print(lista2)
lista3 = lista1 + lista2
print(lista3)

# ordenar la lista de menor a mayor
print(lista1.sort())
print(lista1)

# ordenar la lista de mayor a menor
print(lista1.sort(reverse = True))
# sort lo modifica pero al print devuelve None
print(lista1) # lista ordenada

```
