## Ejercicio 1
Somos una empresa que tiene varias tiendas de electrodomesticos. Necesitamos un programa para manejar las tiendas, ver las ventas que han tenido, cuántos empleados hay, etc... Para ello vamos a modelizar la tienda en una clase, que tendrá las siguientes características:
* Nombre clase: "Tienda"
* Atributos comunes:
    * Tipo: "Electrodomésticos"
    * Abierta: True
* Atributos propios de cada tienda:
    * Nombre - String
    * Dirección - String
    * Número de empleados - int
    * Ventas ultimos 3 meses - Lista de 3 numeros
* Método para calcular las ventas de todos los meses, que devuelva un numero con todas las ventas.
* Método que calcula la media de ventas de los ultimos meses, por empleado
* Método que devuelve en un string el nombre de la tienda, junto con su dirección.
* Método que obtiene las ventas del último mes.
* Método para dar una proyección de las ventas en caso de haber invertido X dinero en marketing en esa tienda. Siendo X el parámetro de entrada. Si X es menor de 1000, las ventas de los ultimos 3 meses hubiesen sido de (1.2 x ventas), si es mayor o igual de 1000, las ventas hubiesen sido de (1.5 x venta). El método reescribe el atributo de ventas en los últimos 3 meses, y además devuelve las ventas con el aumento.

Se pide:
1. Implementar la clase "Tienda"
2. Crear tres tiendas con datos inventados
3. Comprueba en al menos una de ellas, todo lo implementado en la clase tienda (sus atributos, media de ventas, ventas/empleado...)
3. Calcular las ventas del último mes de todas las tiendas. Para ello usa el bucle `for`
4. Imprime por pantalla los nombres de las tiendas cuya dirección lleve el string "Avenida"
5. Seguro que ya has hecho este apartado... Documenta la clase :)

```python
# hacer 2/10/2026 para el 5/10/2026
#1
class Tienda:
    """ 
    Clase con los componentes de una tienda de Electrodomesticos S.A.
    atributos:
    tipo = Electrodomestico --> String
    abierta = True
    ESTO ES UN STRING PARA DOCUMENTAR # añadido en clase 
    # es una buena practica, se puede hacer help()
    # la unica forma de que sea comentario Python no lo lee
    el string multilinea suelto Python si lo lee, ocupa memoria, si es lo ultimo, 
    incluso se puede pedir que se imprima

    """

    tipo = "Electrodomesticos"
    abierta = True

    def __init__(self, nombre, direccion, empleados, ventas): #inicia la tienda y sus datos
        """ Elementos particulares de cada tienda
        Input: 
        nombre: string
        direccion: string
        empleados: int
        ventas: list

        # un metodo es porque esta dentro de una clase aunque sea una funcion __init__

        # aqui nombre es parametro, abajo también, son parametros de instancia
        # self.nombre es un atributo de instancia del self

        # al __init__ no es una buena practica poner return, ya tenemos el self
        no quieres retornar nada estás inicializando

        """

        self.nombre = nombre
        self.direccion = direccion
        self.empleados = empleados
        self.ventas = ventas

    def ventas_totales(self): #apartado 6.  ventas totales
        return sum(self.ventas) #aqui tiene return, en clase, en el __init__ evitar/self

    def media_ventas_empleados(self): #  media de ventas por empleados 
        return sum(self.ventas) / self.empleados

    def nombre_direccion(self):  # nombre de la tienda y su direccion
        return f"{self.nombre} - {self.direccion}"

    def ventas_ultimo_mes(self): # ventas del ultimo mes
        return self.ventas[-1]

    def proyeccion_ventas(self, x): #proyeccion de ventas, segun la inversión de marketing
        if x < 1000:
          self.ventas = [venta * 1.2 for venta in self.ventas]
        else:
          self.ventas = [venta * 1.5 for venta in self.ventas]
        return self.ventas
    
#2
tienda1 = Tienda("Dia", "Calle Lirio 1", 5, [1000, 1200, 1500])
tienda2 = Tienda("Caprabo", "Calle Real 2", 7, [800, 900, 1110])
tienda3 = Tienda("Maravilla", "Avenida Grande 6", 2, [1500, 1800, 2100])

#3
print(tienda1.tipo)
print(tienda1.abierta)
print(tienda1.nombre)
print(tienda1.direccion)
print(tienda1.empleados)
print(tienda1.ventas)


print(tienda1.ventas_totales())
print(tienda1.media_ventas_empleados())
print(tienda1.nombre_direccion())
print(tienda1.ventas_ultimo_mes())

print(tienda1.proyeccion_ventas(500))
print(tienda2.proyeccion_ventas(1000))

#4

tiendas = [tienda1, tienda2, tienda3]
for tienda in tiendas:
    print(tienda.nombre, tienda.ventas_ultimo_mes())

#5
for tienda in tiendas:
    if "Avenida" in tienda.direccion:
        print(tienda.nombre)  # maravilla que es la que tiene direccion Avenida
       
    
```

## Ejercicio 2
En este ejercicio vamos a implementar una clase *Perro* en Python. La clase tiene las siguientes características:
* Cosas que sabemos seguro que tiene un perro
    * Tiene 4 patas
    * 2 orejas
    * 2 ojos
    * Una velocidad de 0. Por defecto, el perro está parado
* Cuando se inicialice:
    * El perro será de una determinada raza
    * Por defecto tendrá pelo "Marrón", a no ser que se diga lo contrario.
    * Por defecto no tendrá dueño, a no ser que se diga lo contrario.
    
* Dispondrá también de un método llamado andar, que tiene un argumento de entrada (aumento_velocidad). Este valor se le sumará a la velocidad que ya llevaba el perro.
* Necesita otro método (parar), donde pondremos la velocidad a 0.
* Otro método llamado "ladrar", que tendrá un argumento de entrada, y la salida será el siguiente string: "GUAU!" + el argumento de entrada.


Se pide:
* Implementa la clase *Perro*
* Crea un objeto de tipo *Perro*, sin dueño
* Comprueba que están bien todos sus atributos
* Prueba que ande, y comprueba su velocidad
* Páralo
* Documenta la clase *Perro*

```python
#hacer 2/10/2026
# __eq__

# def __eq__(self,otra):

# ANOTACIONES EN CLASE. return self.nombre == otra.nombre #otra en el ejemplo objeto de tipo fruta 
# para ejercicio cuenta bancaria

# HACER LOS EJERCICIOS DE LOS ELECTRODOMESTICOS Y EL DE LA CUENTA BANCARIA 


class Perro:
    patas = 4
    orejas = 2
    ojos = 2
    velocidad = 0

    def __init__(self, raza, pelo="Marron", dueño=None):
        self.raza = raza
        self.pelo = pelo
        self.dueño = dueño  # seria mejor dueno, ñ # en ejemplo de clase
        self.velocidad = 0

    def andar(self, aumento_velocidad):
        self.velocidad += aumento_velocidad  # Documentar. Aumenta la velocidad del perro

    def parar(self):
        self.velocidad = 0  # pone la velocidad del perro a 0

    def ladrar (self, mensaje):
        return "GUAU!" + " "  +  mensaje # " " espaciado porque me salia el mensaje junto a GUAU
    
perro1 = Perro("Golden")

print("patas:", perro1.patas)
print("orejas:", perro1.orejas)
print("ojos:", perro1.ojos)
print("velocidad:", perro1.velocidad)
print("raza:", perro1.raza)
print("pelo:", perro1.pelo)
print("dueño:", perro1.dueño)

# comprobacion de que anda y velocidad

perro1.andar(7)
print("Velocidad que tiene tras iniciar a andar:", perro1.velocidad)

# parar al perro

perro1.parar()
print("Velocidad tras parar: ", perro1.velocidad)

# mensaje

print(perro1.ladrar("Quiero salir a la calle"))
```
