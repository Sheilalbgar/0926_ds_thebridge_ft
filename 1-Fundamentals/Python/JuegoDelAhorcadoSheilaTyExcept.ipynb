```python
# AHORCADO 
# practica try/except  14/10/2026

palabra_correcta = "pantalla"

palabra_oculta = ["  _  "] * len(palabra_correcta)
letra_visible =  list(palabra_correcta)

intentos = 7


print("Comienza el juego del Ahorcado")
print("La palabra que buscas es de este tamaño: ")
print(palabra_oculta)

while intentos > 0 and palabra_oculta != letra_visible:
    try:                                                          # try
        letra_input = input("Dime una letra: ").lower()
        if len(letra_input) != 1 or not letra_input.isalpha():    
            raise ValueError("debes escribir una sola letra")     
    except ValueError as e:                                       # except
        print(f"Entrada no válida: {e}")                         
        continue                                                  
    if letra_input in letra_visible:           #parte en el que la letra está en la palabra
        for indice in range(len(letra_visible)):   #buscar la posicion de la letra
            if letra_visible[indice] == letra_input:
                palabra_oculta[indice] = letra_input  #muestra la letra en la palabra oculta
                print(palabra_oculta)
    else:
        intentos = intentos -1 
        print("Error")
        print("Te quedan", intentos, "intentos")
        print(palabra_oculta)
        
if palabra_oculta == letra_visible:  #palabra completada
            print("¡Muy bien!", "¡Ganaste!")
            print("La palabra es: ", palabra_correcta)
        
else:
    print("Perdiste")
    print("La palabra es: ", palabra_correcta)
```
