## Juego del Ahorcado

Planteamiento:

- Proponemos la palabra a adivinar (denominada palabra oculta)
Asignamos a una variable

- Creamos una lista con la longitud igualada a esa palabra propuesta
Para ello utilizamos (len) multiplicando los elementos de esa lista denominados como guiones - por la longitud de la palabra a adivinar.

- Creamos una lista con la palabra en guiones denominada tapada y creamos una segunda lista con la palabra correcta, al convertir la palabra en una lista, nos la muestra separada por palabras.

- A continuación, solicitamos que nos digan una letra, mediante input, al resultado de ese input lo convertimos siempre a minusculas, por si la letra es escrita en mayuscula o minuscula 

Si esa letra se encuentra en la palabra dada (visible), solicitamos que sustituya que en la lista (tapada) la modifique y en vez del guión - muestre en ese indice, en esa posición la letra aportada

Si esa letra no se encuentra en la palabra dada (palabra correcta), solicitamos que lo identifique como error (si error ocurre durante 7 veces, fin del juego, si se consigue destapar la palabra, has ganado)


```python
# planteado en clase

palabra_correcta = "pantalla"

palabra_oculta = ["-"] * len(palabra_correcta)
letra_visible = list(palabra_correcta)
print("tapada", palabra_oculta)
print("visible", letra_visible)
```

```python
# AHORCADO

palabra_correcta = "pantalla"

palabra_oculta = ["-"] * len(palabra_correcta)
letra_visible =  list(palabra_correcta)

vidas = 7

#comienzo
print("Comienza el juego del Ahorcado")
print(palabra_oculta)

while vidas > 0:
    letra_input = input("Dime una letra: ").lower()
    if letra_input in letra_visible:
        for indice in range(len(letra_visible)):
            if letra_visible[indice] == letra_input:
                palabra_oculta[indice] = letra_input
                print(palabra_oculta)
    else:
        vidas < 0
        print("Error")
        print("Te quedan", vidas, "vidas")
        print(palabra_oculta)

        if palabra_oculta == letra_visible:
            print("Ganaste")
            print("La palabra es: ", palabra_correcta)
            break



```
