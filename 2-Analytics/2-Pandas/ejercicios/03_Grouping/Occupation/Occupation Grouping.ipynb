```python
# continuacion

def gender_to_numeric(x):
    if x == "M":
        return 1
    if x == "F":
        return 0

users["gender_n"] = users["gender"].apply(gender_to_numeric)
users
```

```python
# continuacion

users.groupby("occupation").gender_n.sum()

# hace la suma agrupando teniendo en cuenta que siendo hombre vale 1 y  mujer 0
# suma los 1 los hombres porque el 0 no tiene valor no suma 
# y se sabe cual de cada ocupacion es hombre
```

```python
a = (users.groupby("occupation").gender_n.sum() / users.occupation.value_counts() )
     # dividido entre 
a
```
