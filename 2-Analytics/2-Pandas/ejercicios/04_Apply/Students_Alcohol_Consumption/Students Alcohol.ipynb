### Step 5. Create a lambda function that will capitalize strings.

```python
# en la correccion en clase

# lambda palabra reservada de python para hacer funciones
# lambda x,y: x + y
# tenemos x e y y devuelve x + y
# lambda x,y: x + y if x > 10 else None


# lambda x,y: x + y if x > 10 else 20
# se la puede nombrar si despues se la quiere utilizar
# otra_funcion = lambda x,y: x + y if x > 10 else 20
# otra_funcion (2,5) --> devuelvve 20

# funcion anonima la utilzamos por ejempolo cuando queremos hacer...
# en una columna a cada valor le queremos sumar dos df.age +2 lo suma
# si queremos a la columna sumarle dos si el sexo es F y sino le resto 200
# lo hacemos con una funcion
# 

def resta_cosas(x):
    if x ["sex"] == "F":
        return x["age"] + 2
    elif x["sex"] == "M":
        return x["age"] - 200

df.apply(resta_cosas, axis = 1) # --> columnas
# este apply accede columna sex  devuelve una serie

# x es el dataframe completo y aqui trabaja por columnas

# al reves axis = 0 para que funcione una opcion es dt.T trasponerlo y ya podemos
# trabajar con esos datos en axis = 1 por columnas de nuevo para que funcione y tambien apply


# lambda, funcion anonima, no tiene nombre, se puede renombar y utilizar despues

# df.apply(lambda x: x["age"] + 20000 if x["sex"] == "F" else x["age"] - 200, axis =1)

# otra funcion lambda

# list(map(lambda x : str.upper(x), ["hola", "adios"]))  --> map

```

```python

letra_capital = lambda x: x.capitalize()     
df5 = letra_capital
df5("madrid")


# .str.capitalize(): Pone la primera letra en mayúscula, letra capital
# y convierte el resto de los caracteres a minúsculas.

# .str.title(): Pone en mayúscula la primera letra de cada palabra en el texto
```
