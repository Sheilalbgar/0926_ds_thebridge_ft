# JUEGO HUNDIR LA FLOTA


### NORMAS DEL JUEGO
Hundir la flota, es un juego para dos jugadores (o para un jugador contra la máquina). Consta de varios barcos (en este caso de 6) y de dos tableros para cada jugador, en un tablero se colocan los barcos (en este caso de forma aleatoria) y en el otro tablero cada jugador podrá visualizar los barcos que durante el juego y por turnos alternos detecte de su oponente.

Al comenzar el juego los barcos serán posicionados de forma aleatoria, en posición horizontal o vertical, para que cada partida sea diferente. Durante el juego cada jugador "disparará" al tablero del oponente, para intenta localizar los barcos, la forma de realizarlo es eleigiendo una posición de una fila y de una columna, dentro del tablero (como jugador ten en cuenta que tu oponente, la máquina disparará al azar).
Ganará el jugador que antes consiga localizar todos los barcos del oponente, es decir consiga hundir todos los barcos de su rival, haciendo honor al nombre del juego. 

El jugador comenzará la partida y podrá detener el juego en el momento deseado escribiendo, salir.



Ha continuación se desarrolla el código:

```python
import numpy as np

import random

tamano_tablero = 10     # tamaño del tablero

                        # simbolos del tablero
simbolo_agua = "_"      # agua sin disparar
simbolo_barco = "O"     # trozo de barco
simbolo_tocado = "X"    # trozo de barco que ha recibido un disparo
simbolo_fallo = "A"     # disparo que ha caído al agua
```

```python
def crear_tablero(tamano = tamano_tablero):
    return np.full((tamano, tamano), simbolo_agua) #creacion de un tablero de tamaño x relleno de agua ("_")

 
```

```python
def color_barco(barco, tablero):
    for casilla in barco:
        tablero[casilla] = simbolo_barco
        return tablero                                # posicionar en el tablero el barco "O"

```

```python
def crear_barco(eslora, tamano = tamano_tablero):     # posicionar barco al azar con el tamaño indicado
    fila = random.randint(0, tamano -1)
    columna = random.randint(0, tamano -1)
    direcciones = ["N", "S", "E", "O"]                # elige casilla al inicio y orientación al azar (N, S, E, O)
    posicion = random.randint(0, len(direcciones) -1)  # posicion al azar dentro de la lista
    orientacion = direcciones[posicion]

    barco = []
    for _ in range(eslora):
        barco.append((fila, columna))
        if orientacion == "N":
            fila -= 1
        elif orientacion == "S":
            fila += 1
        elif orientacion == "E":
            columna += 1
        else :
            columna -= 1
    return barco

# El barco puede salirse del tablero: comprobar con cabe_barco(). A continuación

```

```python

flota = {2: 3, 3: 2, 4: 1}      # 3 barcos de eslora 2, 2 barcos de eslora 3 y 1 barco de eslora 4
                                # eslora, cuantos barcos hay de esa eslora

def crear_lista_esloras(flota):
    lista = []                  # convertir la flota en una lista con una eslora por barco de MAYOR a menor
    for eslora, cantidad in flota.items():
        for _ in range(cantidad):
            lista.append(eslora)
    lista.sort(reverse=True)    # los barcos grandes primero, por tamaño son los mas dificiles de poner en el panel
    return lista

lista_esloras = crear_lista_esloras(flota)      # [4, 3, 3, 2, 2, 2]  
                                                # 3 barcos de eslora 2, 2 barcos de eslora 3 y 1 barco de eslora 4

```

```python
def cabe_barco(barco, tablero):         # es True si todas las casillas del barco estan dentro del tablero y libres
    tamano = tablero.shape[0]
    for fila, columna in barco:
        if fila < 0 or fila >= tamano or columna < 0 or columna >= tamano:
            return False                # si se sale del tablero
        if tablero[fila, columna] != simbolo_agua:
            return False                # si ya hay otro barco
    return True



def colocar_barcos(tablero, esloras = lista_esloras): # colocar al azar los barcos, sin solaparse ni salir del tablero
    tamano = tablero.shape[0]
    for eslora in esloras:
        while True:
            barco = crear_barco(eslora, tamano)
            if cabe_barco(barco, tablero):
                colocar_barcos(barco, tablero)
                break                                 # si el barco ya está, pasa al siguiente
    return tablero

```

```python

def disparos(casilla, tablero):
    if tablero[casilla] == simbolo_barco:
        tablero[casilla] = simbolo_tocado
        return "tocado"
    elif tablero[casilla] == simbolo_agua:
        tablero[casilla] == simbolo_fallo
        return "agua" 
    else:
        return "repetido"          
          
```

```python

def quedan_barcos(tablero):
    #True si queda alguna casilla de barco sin tocar.
    return np.any(tablero == simbolo_barco)


def disparo_maquina(disparos_maquina):
    #La máquina elige al azar una casilla a la que todavía no ha disparado.
    libres = np.argwhere(disparos_maquina == simbolo_agua)   # casillas sin disparar
    posicion = random.randint(0, len(libres) - 1)           # posición al azar dentro de esa lista
    fila, columna = libres[posicion]
    return (int(fila), int(columna))


```
