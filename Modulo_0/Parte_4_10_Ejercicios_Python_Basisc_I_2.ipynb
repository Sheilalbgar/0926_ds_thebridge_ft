## Ejercicio 2
Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

radianes = grados*(pi/180)

```python
# radianes = grados * (pi/180)
# se importa el modulo math que tiene el valor de pi
# y se convierte el input a float (numero real)
```

```python
import math

grados = float(input("Introduce los grados: "))

radianes = grados * (math.pi / 180)

print(grados, "grados son", radianes, "radianes")
```

## Ejercicio 3

## Ejercicio 4
Escribe un programa que calcule el area de un paralelogramo (base x altura). También con `input`. En este caso debe ser capaz de admitir valores reales, con decimales, para base y altura en español. (PISTA: Recuerda que tendrás que hacer uso del método replace)

```python
# igual que el ejercicio anterior que admite decimales
#  pero corrigiendo que en la escritura del input el usuario pueda introducir el decimal con , coma o con punto .

base = input("Introduce la base: ")
altura = input("Introduce la altura: ")

base = float(base.replace(",", "."))
altura = float(altura.replace(",", "."))

area = base * altura

print("El área del paralelogramo es", area)


```
